## 1. Interactive vs. script <a class="anchor" id="sec1"></a>

## 2. Functions <a class="anchor" id="sec2"></a> 

Once a piece of code is worth reusing, wrap it in a **function**. A function is a mini-script that (optionally) takes some inputs (`arguments`) and (optionally) returns a value.

```python
def moffat1D(r, I, alpha, beta):
    arg = (r / alpha)**2 + 1
    y = I / arg**(beta)
    return y
```

This evaluates a **Moffat profile** (common in astronomy to model the point-spread function of a star image):

$y = I_0 \left( \left( \dfrac{r}{\alpha} \right)^2 + 1 \right)^{-\beta}$

**The rules that matter:**
- `def name(...):` starts the function; the indented block below it is the function body.
- A variable created inside the function (like `arg` above) is **local** — it does not exist outside the function.
- The function ends at `return`, or when the indentation goes back to the outer level.
- **Argument order:** mandatory arguments (no default value) first, arguments with a default value after — you cannot put a mandatory one after a default one.
- An argument with a default value is *optional* when calling the function: you can also override it by name, e.g. `moffat1D(2, 2, alpha=3, beta=1)`.


````python
def moffat1D_no_default(r, I0, alpha, beta):
    """
    I0 = Intensity at origin
    """
    arg = (r / alpha)**2 + 1
    y = I0 / arg**(beta)
    return y

moffat1D_no_default(2, 2, 3, 1)
````

## 3. Functions and docstrings — do this every time <a class="anchor" id="sec3"></a>

A **docstring** is a short piece of documentation placed right inside the function, between triple quotes `""" ... """` as the very first line of the body. It's what shows up when you call `help(my_function)`, or in Jupyter with `Shift+Tab` after typing the function name, or with `my_function?`.

**This is not optional decoration.** A function without a docstring is, in practice, an unfinished function: nobody (including future-you, in three weeks) can tell what it does, what it expects, or what it returns without re-reading, and possibly re-understanding, the whole implementation. **From this point on in the course, every function you write should have a docstring.**

A widely used format (NumPy-style, also readable by documentation tools like Sphinx):

```python
def moffat1D(r, I, alpha, beta):
    """
    Evaluate a Moffat profile at position r.

    Parameters
    ----------
    r : float
        Position at which the function is evaluated.
    I : float
        Intensity (normalization).
    alpha : float
        Scale parameter (width) of the profile.
    beta : float
        Shape parameter of the profile.

    Returns
    -------
    float
        The value of the Moffat function at position r.
    """
    arg = (r / alpha)**2 + 1
    y = I / arg**(beta)
    return y
```

You don't have to follow this format to the letter every time (a one-line description is fine for a short, simple function) but the essentials (i.e., what it does, what goes in, what comes out) should always be there. See the [docstring convention (PEP 257)](https://www.python.org/dev/peps/pep-0257/) and the [numpydoc format](https://numpydoc.readthedocs.io/en/latest/format.html#format) if you want the full guidelines.

### `*args` and `**kwargs` (optional, for later reference)

*Come back to this section once you're comfortable with functions — you don't need it to get started, but you'll run into this notation reading other people's code.*

Sometimes a function is defined with `*args` or `**kwargs` instead of an explicit list of parameters:

```python
def line(x, a, b):
    return [a * xi + b for xi in x]

def line_args(x, *args):        # *args packs all extra positional arguments into a tuple
    a, b = args
    return [a * xi + b for xi in x]

def line_kwargs(**kwargs):      # **kwargs packs all keyword arguments into a dict
    return [kwargs['a'] * xi + kwargs['b'] for xi in kwargs['x']]
```

The `*` and `**` are **unpacking operators**: `*` unpacks an iterable (tuple/list) into positional arguments, `**` unpacks a dictionary into keyword arguments — this works both when *defining* a function (to collect arbitrary arguments) and when *calling* one (to spread out a tuple/dict you already have):


````python
def line(x, a, b):
    return [a * xi + b for xi in x]

x = [0, 1, 2, 3, 4]

params = (0.5, 1)          # a tuple of the remaining arguments
print(line(x, *params))    # unpacked when calling

kw = {'a': 0.5, 'b': 1}    # a dict of keyword arguments
print(line(x, **kw))       # unpacked when calling
````

One rule to remember: if a function uses both, the order is always `def f(arg1, arg2, ..., *args, **kwargs):`.

### A. Interactive python outside of a notebook

Outside a notebook, one can use the standard Python prompt, which can be launched by typing `python` in a terminal. Instead, one can use a feature enhanced prompt, `ipython`, where "i" stands for interactive (which also means that IPython is an *interpreter*). This is an add-on package that adds many features to the default Python shell, including the ability to edit and navigate the history of previous commands, as well as the ability to tab-complete variable and function names. You can also write several commands one after the other before executing them by typing `Ctrl Enter` between each command instead of `Enter`. For some versions of ipython, this is rather `Ctrl o`. `IPython` allows you also to use any shell command by preceding it with a `!` as we did at the beginning of this Notebook (If this does not work, try without the `!`). This is basically telling Ipython that this shell is activated in the *code* cells of this Jupyter Notebook.    

Once you have launched `ipython`, you can proceed as follows to type your `python` commands: 
``` python

```

When you use an `ipython` shell, you can access the command history by pressing up/down arrows. This feature doesn't work within jupyter.

## Summary <a class="anchor" id="summary"></a>

- Interactive mode (incl. Jupyter) for exploring; scripts (`.py` files) for anything you want to reproduce or share.
- A function packages reusable code: `def name(args): ... return value`. Mandatory arguments first, arguments with defaults after.
- **Write a docstring for every function, every time.** It's the difference between reusable code and code only you can (temporarily) understand.
- `*args`/`**kwargs` let a function accept a flexible number of positional/keyword arguments: good to recognize when reading code, less essential to write yourself right away.
