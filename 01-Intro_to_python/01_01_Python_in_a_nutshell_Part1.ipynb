## 1. Python vs. what you already know <a class="anchor" id="sec1"></a>

You already know the mental model of programming: variables, control flow, functions. What's actually new for you is *Python's specific way of doing it*. So instead of rebuilding programming from scratch, this notebook is organized around **the deltas**: what's different from C/C++/Java, and where those differences bite people the first time.


### 1.1 Interpreted, and why that matters for you <a class="anchor" id="sec11"></a>

| | C / C++ / Java | Python |
|---|---|---|
| Workflow | write &rarr; compile &rarr; run | write &rarr; run (compiled to bytecode on the fly, then interpreted) |
| Type checking | at compile time | at run time (dynamic typing) |
| Variable declaration | required (`int x;`) | none — a name is just bound to a value |
| Typical use | build & ship a program | both scripts **and** an interactive shell for exploration |

The practical consequence: you can type a line of Python and see the result immediately, without a build step. This is why Python is so heavily used for exploratory / interactive work (data analysis, prototyping) in addition to "real" programs. We'll use this interactive style constantly in this course, in a Jupyter notebook exactly like this one.

**No semicolons, no braces.** Statement ends = end of line. Blocks (the body of an `if`, a `for`, a function, ...) are **not** delimited by `{ }` — they are delimited by **indentation**. This is not just a style convention, it's syntax: inconsistent indentation is a `SyntaxError` (technically an `IndentationError`) or, worse, silently changes what your code does.


````python
# Run this cell (Shift+Enter)
print("Hello")
````

````python
# This also works on its own, without print, in a notebook / interactive shell
"Hello world!"
````

### 1.2 First steps: no declarations, indentation is syntax <a class="anchor" id="sec12"></a>

No type, no declaration keyword: a name is bound to a value with `=`, and that's it.


````python
x = 10
x       # in a notebook / REPL, the value of the last line is shown automatically
````

`Python` is **case-sensitive**: `x` and `X` are two different names, exactly like in C/Java.

Printing several things at once: separate them with commas in `print(...)`.


````python
print('Result =', x / 2)

# printf-style formatting also exists (more on string formatting later):
print('Result = %.2f' % (x / 2))
````

## 2. Core types of python variable <a class="anchor" id="sec2"></a>

### 2.1 Numbers, booleans, `None` <a class="anchor" id="sec21"></a>

Like `int`/`double`/`boolean`/`null` in Java / C, but nothing to declare, and the type is attached to the *value*, not to the variable name. Use the built-in `type()` to check it (Modify the next cell by yourself to see it).


````python
a = 2          # no '.' -> int
b = 2.0        # '.'    -> float
c = True       # bool
d = None       # NoneType: "no value" -- NOT the same as 0, False, or ""

for v in (a, b, c, d):
    print(v, '->', type(v))
````

`None` is Python's `null`, with one important difference: only `None` is `None`. You never check equality with `==`, you check *identity* with `is` (we'll see why `is` exists at all in a moment).


````python
n = None
print(n is None)        # idiomatic
print(n == None)        # works, but not idiomatic -- avoid it
````

**Exercise 2 - `is` vs `==`.** 

Predict, then check: are these two booleans referring to the *same* object, or just *equal* values? Create `p = 1000` and `q = 1000`.   
Compare `p == q` and `p is q`. Then try `p = 5` and `q = 5` and compare again. (Python caches small integers, so the result differs — this is exactly the kind of surprise that `is`-vs-`==` produces if you assume it behaves like value equality everywhere.).  
Rule of thumb: **use `==` for *value* comparison, always** ; reserve `is` for `None`/singleton checks.

````python
# (2) is vs == experiment

p = 1000
q = 1000
print(p == q)           # True, same value
print(p is q)           # False, different objects
````

### 2.2 Operators: the ones that differ from C/Java <a class="anchor" id="sec22"></a>

`+ - * /` work as expected. Three operators do **not** exist (in this form) in C/Java:

| Operator | Name | Example | Result |
|---|---|---|---|
| `**` | power | `2 ** 10` | `1024` |
| `//` | floor (integer) division | `7 // 2` | `3` |
| `%` | modulus | `7 % 2` | `1` |

Also note: dividing two `int`s with `/` in Python 3 **always returns a `float`** (unlike C/Java, where `int / int` truncates).


````python
print(7 / 2)     # float division -> 3.5
print(7 // 2)    # floor division -> 3
print(7 % 2)     # modulus -> 1
print(2 ** 10)   # power -> 1024
````

Comparison operators (`== != < > <= >=`) work exactly as in C/Java. One genuine novelty: Python lets you **chain** comparisons.


````python
x = 5
print(1 < x < 10)      # equivalent to (1 < x) and (x < 10) -- but Python lets you write it directly
````

### A.1 A (very) brief history

`Python` was created by the Dutch programmer Guido van Rossum, conceptualized in the late 1980s while he worked on Amoeba, a distributed operating system, and released publicly in 1991. The name comes from ([Monty Python's Flying Circus](https://en.wikipedia.org/wiki/Monty_Python%27s_Flying_Circus)), not the snake. 

Python 2 and Python 3 co-existed for a long time (Python 2 reached end-of-life in 2020) — if you ever see `print "text"` without parentheses in old code, that's Python 2 syntax.

### A.2 Complex numbers

Python has a built-in `complex` type — useful in a few scientific contexts, less so for general-purpose programming.


````python
c = 4 + 7j          # note: no '*' before the imaginary unit 'j'
c1 = complex(4, 7)   # equivalent, explicit constructor
print(type(c), c, c1)
````

### A.3 Exhautive list of operators 

Here is a more complete list of operators
In the following examples, let us take `a=10`, `b=20`.

|Operator 	| Name | Description |	Example   |
|-----------|-------|------|-----------|
| `+` | Addition 	| Adds values on either side of the operator. | 	`a + b = 30`  |
| `-` | Subtraction |	Subtracts right hand operand from left hand operand. |	`a – b = -10` | 
| `*` | Multiplication |	Multiplies values on either side of the operator  |	`a * b = 200` | 
| `/` | Division 	| Divides left hand operand by right hand operand |	`b / a = 2` |
| `%` | Modulus 	| Divides left hand operand by right hand operand and returns remainder |	`b % a = 0` | 
| `**` | Exponent |	Performs exponential (power) calculation on operators |	`a**b =` $10^{20}$ |
| `//` | Floor Division | The division of operands where the result is the quotient in which the digits after the decimal point are removed. But if one of the operands is negative, the result is floored, i.e., rounded away from zero (towards negative infinity) | 	`9//2 = 4` and `9.0//2.0 = 4.0`, `-11//3 = -4`, `-11.0//3 = -4.0` |

Comparison operators are like in `C` programming language.    


|Operator 	| Description |	Example |
|-----------|-------------|---------|
|`==` | 	If the values of two operands are equal, then the condition becomes true. |	`(a == b)` is False |
| `!=` |	If values of two operands are not equal, then condition becomes true. 	| `(a != b)` is True. |
| `>` 	| If the value of left operand is greater than the value of right operand, then condition becomes true. | 	`(a > b)` is False. | 
| `<` 	| If the value of left operand is less than the value of right operand, then condition becomes true. 	| `(a < b)` is True. | 
| `>=` 	| If the value of left operand is greater than or equal to the value of right operand, then condition becomes true.| 	`(a >= b)` is False. | 
| `<= `|	If the value of left operand is less than or equal to the value of right operand, then condition becomes true. 	| `(a <= b)` is True.| 


## Summary <a class="anchor" id="summary"></a>

- Python is interpreted (compiled to bytecode on the fly), which is why it doubles as a scripting language *and* an interactive shell.
- No variable declarations; a name is bound to a value with `=`. Indentation is syntax, not style.
- Core scalar types: `int`, `float`, `bool`, `None` (and `complex`, less commonly used). Check types at runtime with `type()`.
- `==` for value equality (always); `is` only for `None` / identity checks.
- New-to-you operators: `**` (power), `//` (floor division), `%` (modulus). `int / int` always returns a `float`.
