### 1.1 Lists and slicing <a class="anchor" id="sec11"></a>

A list is an ordered, mutable sequence (think `ArrayList<Object>` in Java ; `std::list` in C++), but built into the python language and able to mix types. Indexing is **zero-based**, exactly like arrays in C/Java.


````python
L = [10, 20, 30, 40, 50]
print(L[0])     # first element
print(L[-1])    # last element -- negative indices count from the end
````

**Slicing** (`L[start:stop:step]`) is genuinely new: it lets you pull out a sub-list without a loop. `stop` is *exclusive*.

```python
L[0:2]     # elements at index 0 and 1 (NOT 2)
L[2:]      # from index 2 to the end
L[:3]      # from the start up to (not including) index 3
L[::2]     # every 2nd element
L[::-1]    # the whole list, reversed
```


````python
# (1) Slicing basics: using L = [10, 20, 30, 40, 50] above, get:
#   - the first three elements
#   - the last two elements
#   - the list reversed

L = [10, 20, 30, 40, 50]
print('First two elements', L[0:2])
print('Last two elements', L[-2:])
print('List Reversed', L[::-1])
print('Last two elements', L[-2:16]) # another way to go until the end: end value > len(L)
````

````python
print(len(L))
````

````python
# (2) Slicing with a step: get every 2nd element of L, starting from index 1

print('Every two elements starting from 1', L[1::2])
````

Slicing also works on strings (a string is a sequence too, just an *immutable* one).


````python
# (3) Take the string s = "Programming" and extract, using slicing only (no loop):
#   - the first 4 letters
#   - the last 4 letters
#   - every other letter
s = "Programming"

print('First 4 letters', s[:4])
print('the last 4 letters', s[-4:])
print('Every other letter', s[::2])
print(len(s))
````

### 2.1 `if` / `elif` / `else` <a class="anchor" id="sec21"></a>

In "C", one has the following structure: 
``` c
if (a == 1) {
    printf("a is 1");
} else if (a == 2) {
    printf("a is 2");
} else {
    printf("other");
}
```

In `Python`, WE HAVE: 

``` python 
if a == 1:
    print("a is 1")

else:
    print("other")
``` 


````python
a = 1
if a == 1:
    print("a is 1")
elif a == 2:
    print("a is 2")
    print("strange, strange")
else:
    print("other")
    print("even more strange")
````

Truthiness matters: any container is `False` in a boolean context when empty, `True` otherwise.    
You do not need to compare against `len(...) == 0` to know if a container contains something.


````python
for container in ([], [False], {}, {0: 'a'}, set(), (1,)):
    if container:
        print(container, '-> truthy')
    else:
        print(container, '-> falsy')
````

````python
# Short digression 1/3 (while we are at strange things)
# In the tuple above, the last element is a one-element tuple
# What would be the result if the comma is left out?
# Answer: type(x) would be int, as the brackets act as an arithmetic group delimiter,
#         not as a tuple definition character. The comma distinguishes its contents from arithmetic operation)
x = (1,)
print(x)
type(x)
````

````python
# Short digression 2/3 (while we are at strange things)
# Python's integer type does not have any value limitation: the sky (your computer memory) is the limit...
print(2**2000)
````

````python
# Short digression 3/3: how to split long lines in Python

# Possibility 1 - works always: end a line that is meant to be continued
# on the following line by a backslash ('\'), which means that it should be
# concatenated with that following
L = [1, 2, 34, 56, \
    356, 78, 'rty', 'not', \
    34, 'h']
print("L  =", L)

# Possibility 2 - works for lines that span the contents of brackets -- (...), [...], {...}.
# No need (but not harm either) for backslashes at the end of lines that leave
# a given type of parenthesis open, until the line that closes it  

L1 = [1, 2, 34, 56,          # The square bracket opened on the current line is closed on the
    356, 78, 'rty', 'not',   # line below. The code on the lines holding the opening and the
    34, 'h']                 # corresponding closing bracket is considered to be a single command.

print("L1 =", L1)
````

### 2.2 `for` loops: `range`, `enumerate`, `zip` <a class="anchor" id="sec22"></a>

In `C`, you have:  
```c
for (int i = 0; i < 5; i++) {
    printf("%d\n", i);
}
```

The direct translation uses `range`:


````python
# range(n) is an object that only contains the minimum information required
# to produce what it means: the sequence of all integers from 0 to n-1.
# But, unlike a list [0, ..., n-1], it does not need to be expanded in memory to be used.
# It can therefore handle range(10**15), which the list [0, ..., 10**15-1]  could not
# (lack of memory).

for i in range(5):
    print(i)
````

````python
# range(i, e): i, i+1, ..., e-1 (range from i to e-1)
# Corollary: range (e) is equivalent to range(0, e)
for i in range(4, 10):
    print(i)
````

````python
# range(i, e, s): i, i+s, ..., es, where es = i + n*s and es < s <= es+s
# Accordingly, the sequence 0, 3, 6, ..., (e//s)*s-1 must be range(0, e, s)
# (i.e., the first argument '0' is required in this case, so that the two other
# arguments are interpreted as required.
for i in range(4, 10, 3):
    print(i)
````

````python
# range(i, e, s): empty if i > e and s > 0
for i in range(21, 15, 3):
    print(i)
````

````python
# range(i, e, s): not empty if i > e and s < 0
# In this case the resulting sequence is i, i+s, ..., es, where es = i + n*s and es-s <= s < es
for i in range(21, 15, -3):
    print(i)
````

But the more *idiomatic* Python is to loop directly over the collection, without an index at all, when you don't actually need the index:


````python
L_of_int_and_str = [1, 4, 'aba', 'cbd', 8]

for item in L_of_int_and_str:
    print(item)
````

## 3. List comprehensions <a class="anchor" id="sec3"></a>

This is the one construct in this notebook that has **no direct C/Java equivalent** (Java streams get close, but the syntax and mindset are different).
**Fortran** has a similar construct (called *implied do-loop*, that is commonly used to initialize array values).

A list comprehension builds a list in a single expression instead of a `for` loop + `.append()`.


````python
# The for-loop way you'd instinctively write:
l = []
for i in range(1, 10):
    l.append(i ** 2)
print(l)
````

````python
# The equivalent list comprehension:
l2 = [i ** 2 for i in range(1, 10) if i % 2 == 1]
print(l2)
````

Read `[EXPRESSION for VARIABLE in ITERABLE]` left to right as *"the expression, for each variable in the iterable"*. You can also add a filter condition at the end.

```python
[expr for x in iterable]              # basic
[expr for x in iterable if condition] # with a filter
```

When possible, prefer list comprehensions over explicit `for` + `append` loops. They're shorter, and considered more "pythonic".

**Boolean vs elementwise operators: `and`/`or` vs `&`/`|`:** 

`and`/`or` are boolean operators. They operate on single `True`/`False` values. 
`&`/`|` are bitwise/elementwise operators. 

Trying to use `and`/`or` element-by-element on two lists does not do what you might expect from array languages. This becomes especially relevant once we start using NumPy arrays later in the course.


````python
mylist1 = [True,  True,  True,  False, True]
mylist2 = [False, True, False, True,  False]

print("with the 'and' operator:", mylist1 and mylist2)   # boolean 'and' on lists: NOT elementwise -- just returns one of the two lists

# elementwise boolean comparison of lists: for loop
mylist1and2 = []
for i in range(len(mylist1)):
    mylist1and2.append(mylist1[i] and mylist2[i])

print("with 'for' loop........:", mylist1and2)

# elementwise boolean comparison of lists: list comprehension
mylist1and2_listcomp = [item1 and item2 for (item1, item2) in zip(mylist1, mylist2)]

print('By list comprehension..:', mylist1and2_listcomp)
````

**Why is `mylist1 and mylist2` equal to one of the two lists? and which one on the end?**

The explanation has two parts:
 1. in the expression `mylist1 and mylist2`, the two operands `mylist1` and `mylist2` are expected to be boolean. If they are not actually boolean, they are *somehow* interpreted as such.
 2. a pecularity of the way `and` is evaluated.

We have seen above ("Truthiness matters...", in Sect. 2.1), that non-empty container structures (i.e., lists, sets, tuples, ...) evaluate to True if they are not empty (whatevver their contents), and to False if they *are* empty. Accordingly, in the example above, `mylist1` is evaluated as True when interpreted as a boolean.

The evaluation of `a and b` works as follows in practice: `a` is evaluated first.
 - If it is True, the result of `a and b` is True if `b` is True, and it is False if `b` is False. Accordingly, `a and b` reduces to `b` in this case.
 - If it is False, `a and b` is False, whatever the result of the evaluation of `b`. Accordingly, `a and b` reduces to `a` in this case.

Since `mylist1` is not empty, `mylist1` interpreted as (fake) boolean is True, and the result of `mylist1 and mylist2` is thus `mylist2`, as explained before.

NB: similar reasonging can be applied to the `or` operator.
