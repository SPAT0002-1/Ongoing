## 1. Containers <a class="anchor" id="sec1"></a>

The built-in containers is where Python actually diverges from C/Java. Tbere are four built-in containers that cover almost everything: **lists**, **tuples**, **sets**, **dictionaries**.

### 1.1 Lists and slicing <a class="anchor" id="sec11"></a>

A list is an ordered, mutable sequence (think `ArrayList<Object>` in Java ; `std::list` in C++), but built into the python language and able to mix types. Indexing is **zero-based**, exactly like arrays in C/Java.


````python
L = [10, 20, 30, 40, 50]
print(L[0])     # first element
print(L[-1])    # last element -- negative indices count from the end
````

**Slicing** (`L[start:stop:step]`) is genuinely new: it lets you pull out a sub-list without a loop. `stop` is *exclusive*.

```python
L[0:2]     # elements at index 0 and 1 (NOT 2)
L[2:]      # from index 2 to the end
L[:3]      # from the start up to (not including) index 3
L[::2]     # every 2nd element
L[::-1]    # the whole list, reversed
```


### 1.2 Tuples: the `const` you never had <a class="anchor" id="sec12"></a>

A tuple `(...)` behaves like a list but is **immutable**: once created, it cannot be changed. Use it whenever a collection of values should never be modified, e.g. to guarantee that a function doesn't accidentally mutate data you pass it.

A **dictionary** stores unordered sequence(s) of key-value pairs. It is defined using curly brackets `{ }` and `key: value` syntax (-> `dict = {key1: value1, key2: vlaue2}` with mixed values types allowed).

## 2. Control flow <a class="anchor" id="sec2"></a>

The *concepts* here (branching, loops) are not new to you. Only the syntax is. So this section is deliberately syntax-first: for each construct, we show the C/Java version you already know, then the Python equivalent.

### 2.1 `if` / `elif` / `else` <a class="anchor" id="sec21"></a>

In "C", one has the following structure: 
``` c
if (a == 1) {
    printf("a is 1");
} else if (a == 2) {
    printf("a is 2");
} else {
    printf("other");
}
```

In `Python`, WE HAVE: 

``` python 
if a == 1:
    print("a is 1")
elif a == 2:
    print("a is 2")
else:
    print("other")
``` 


````python
a = 3
if a == 1:
    print("a is 1")
elif a == 2:
    print("a is 2")
else:
    print("other")
````

Truthiness matters: any container is `False` in a boolean context when empty, `True` otherwise.    
You do not need to compare against `len(...) == 0` to know if a container contains something.


````python
for container in ([], [False], {}, {0: 'a'}, set(), (1,)):
    if container:
        print(container, '-> truthy')
    else:
        print(container, '-> falsy')
````

And, as seen in Part 1: use `is` (not `==`) to test against `None`.

### 2.2 `for` loops: `range`, `enumerate`, `zip` <a class="anchor" id="sec22"></a>

In `C`, you have:  
```c
for (int i = 0; i < 5; i++) {
    printf("%d\n", i);
}
```

The direct translation uses `range`:


````python
for i in range(5):
    print(i)
````

But the more *idiomatic* Python is to loop directly over the collection, without an index at all, when you don't actually need the index:


````python
L_of_i_and_s = [1, 4, 'aba', 'cbd', 8]

for item in L_of_i_and_s:
    print(item)
````

To iterate over two sequences in parallel (like walking two arrays with the same index in C), use `zip` instead of indexing both by hand:


````python
names = ['Ada', 'Alan', 'Grace']
scores = [98, 87, 93]

for name, score in zip(names, scores):
    print(name, '->', score)
````

### 2.3 `while` loops <a class="anchor" id="sec23"></a>

Identical concept and near-identical syntax to C/Java, just without parentheses around the condition and without braces.

In `C`: 
```c
int a = 0;
while (a < 10) {
    printf("%d\n", a);
    a++;
}
```
In `Python`: 
```python
while a < 10:
    print(a)
    a += 1     # no ++ operator in Python
``` 

````python
a = 0
while a < 10:
    print(a)
    a += 1     # no ++ operator in Python
````

## 3. List comprehensions <a class="anchor" id="sec3"></a>

This is the one construct in this notebook that has **no direct C/Java equivalent** (Java streams get close, but the syntax and mindset are different).

A list comprehension builds a list in a single expression instead of a `for` loop + `.append()`.


````python
# The for-loop way you'd instinctively write:
l = []
for i in range(1, 10):
    l.append(i ** 2)
print(l)
````

````python
# The equivalent list comprehension:
l2 = [i ** 2 for i in range(1, 10)]
print(l2)
````

Read `[EXPRESSION for VARIABLE in ITERABLE]` left to right as *"the expression, for each variable in the iterable"*. You can also add a filter condition at the end.

```python
[expr for x in iterable]              # basic
[expr for x in iterable if condition] # with a filter
```

When possible, prefer list comprehensions over explicit `for` + `append` loops. They're shorter, and considered more "pythonic".

**Boolean vs elementwise operators: `and`/`or` vs `&`/`|`:** 

`and`/`or` are boolean operators. They operate on single `True`/`False` values. 
`&`/`|` are bitwise/elementwise operators. 

Trying to use `and`/`or` element-by-element on two lists does not do what you might expect from array languages. This becomes especially relevant once we start using NumPy arrays later in the course.


````python
mylist1 = [True,  True,  True,  False, True]
mylist2 = [False, True, False, True,  False]

print(mylist1 and mylist2)   # boolean 'and' on lists: NOT elementwise -- just returns one of the two lists
````
