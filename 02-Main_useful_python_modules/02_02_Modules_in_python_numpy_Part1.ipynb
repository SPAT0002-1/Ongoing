## II.1 `array` definition and construction <a class="anchor" id="II.1"></a>

A `numpy` array is an object of type `np.ndarray`. There are several ways to build one:

``` python
import numpy as np
np.array([1, 2, 3, 4])              # from a python list
np.array([[0, 1, 2], [3, 4, 5]])    # 2D, from a nested list
np.empty(shape=(2, 3))              # uninitialized array, 2 rows x 3 columns
np.arange(5)                        # like the built-in range()
np.linspace(1, 10, 10)              # 10 evenly spaced elements from 1 to 10
np.zeros(10)                        # 10 elements, all 0
np.ones(10)                         # 10 elements, all 1
np.zeros((2, 5))                    # 2D, 2 rows x 5 columns
```

2D arrays of `shape=(r, c)` have `r` **rows** and `c` **columns**.


````python
import numpy as np
a = np.array([[1, 22, 3], [3, 5, 5]])
a
````

````python
zero_array = np.zeros(shape=(2, 3))
ones_array = np.ones(shape=(2, 2, 3))
print(zero_array)
print(ones_array.shape)   # note: 3 dimensions now
````

````python
print(np.arange(0., 5., 0.5))   # start, stop (excl.), step
print(np.linspace(0, 5, 9))     # start, stop (incl.), number of points
````

`numpy` can also generate random arrays:
```python
np.random.random(size=4)   # uniform between 0 and 1
np.random.normal(size=4)   # standard-normal distributed
```
and compute common math functions elementwise: `np.log10()`, `np.log()`, `np.exp()`, `np.sin()`, `np.cos()`, etc. (full list [here](https://docs.scipy.org/doc/numpy/reference/routines.math.html)).


````python
# try a couple of the commands above
print(np.random.random(size=4))
print(np.log(2.3))
````

## Intermezo: python is an object-oriented language <a class="anchor" id="intermezo"></a>

Creating a variable of a given type generally means creating an **object**. You don't need deep OOP knowledge for this course, but three words will come up constantly:

- **instance**: `a = np.array([[1, 2, 3], [3, 5, 5]])` creates an *instance* of the array object (in practice: allocates memory for it).
- **attribute**: a property attached to an instance, e.g. `a.shape`. Access it with `instance.attribute` (no parentheses).
- **method**: a function attached to an instance that can act on it, e.g. `a.mean()`. Access it with `instance.method(args)` (parentheses, even with no arguments).

Most attributes/methods also have an equivalent standalone function: `a.shape` &harr; `np.shape(a)`, and `a.mean()` &harr; `np.mean(a)`.


````python
print(a.shape)        # attribute -- no parentheses
print(np.shape(a))     # equivalent standalone function
print(a.mean())        # method -- parentheses, even with no arguments
````

## II.1bis Data types (`dtype`) <a class="anchor" id="dtype"></a>

Every array has a `dtype`: the type of the elements it contains (as opposed to `type(a)`, which just tells you `a` is a `numpy.ndarray`). The default is floating point; you can also request `int`, `complex`, `bool`, or string types explicitly.

```python
c = np.array([1, 2, 3], dtype=float)       # dtype('float64')
d = np.array([1+2j, 3+4j, 5+6j])           # dtype('complex128')
e = np.array([True, False, True])          # dtype('bool')
f = np.array(['abc', 'eddafg', 'hjk'])     # dtype('<U6') -- strings, max length 6
```

**Important:** only *arrays* have a `.dtype`. A plain python `list` doesn't (`L.dtype` raises an `AttributeError`) — `type()` works on both, `.dtype` is array-only.
